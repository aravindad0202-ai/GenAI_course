# single shot prompting

````python
prompt = """You are an expert clinical data extractor. Your task is to extract patient data from rough clinical notes into a strict JSON format.

EXAMPLE:
Input note: "Pt is a 45yo male, John Doe, presenting with acute migraines. BP 120/80. Prescribed Sumatriptan 50mg PRN. No known allergies. Follow up in 2 weeks."
Output JSON:
{
  "patient": {"name": "John Doe", "age": 45, "gender": "male"},
  "vitals": {"blood_pressure": "120/80"},
  "clinical": {"diagnosis": ["acute migraines"], "prescriptions": [{"drug": "Sumatriptan", "dosage": "50mg", "frequency": "PRN"}], "allergies": "None"},
  "follow_up_days": 14
}

YOUR TASK:
Input note: "62 yr old female patient Sarah Jenkins. Complains of chronic joint pain in left knee. BP is 135/88, HR 72. Current meds: Ibuprofen 400mg daily. Allergic to Penicillin. Scheduled for MRI on Oct 12th."
Output JSON:"""

response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )
print(response.text)
````

# Few shot pompting

````python
prompt = """Analyze the corporate PR statement and identify the primary logical fallacy being used to deflect blame. 

Statement: "While critics point to our environmental record, we should really be looking at the massive pollution caused by the agricultural sector."
Fallacy: Whataboutism (Tu Quoque) - Deflecting criticism by pointing to the flaws of another, unrelated party.

Statement: "If we don't mandate 80-hour work weeks, our competitors will crush us, we will go bankrupt, and everyone will lose their jobs and homes."
Fallacy: Slippery Slope - Arguing that a relatively small first step inevitably leads to a chain of related (and catastrophic) events.

Statement: "Our new data privacy policy is the best in the industry because it perfectly secures user data."
Fallacy: Circular Reasoning - The conclusion is included in the premise; claiming it is the best because it perfectly does the thing it is supposed to do.

Statement: "People complaining about our new pricing tier clearly just don't understand basic economics or the free market."
Fallacy:"""

response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

print(response.text)
````

# Chain of thoughts prompting

````python
prompt = """Determine if it is logistically possible to fulfill Order Alpha today. 
Do not give me the final answer immediately. First, think through the problem step-by-step, listing your calculations and logical deductions. End your response with "FINAL CONCLUSION: [Yes/No]".

Order Alpha parameters:
- Requires shipping 4,500 units of Product X.
- Must leave the warehouse by 16:00 today.

Warehouse constraints:
- We currently have 5,000 units of Product X in stock.
- We have 2 trucks available. Truck A holds 2,000 units. Truck B holds 3,000 units.
- Truck A is currently out on delivery and will return at 14:30. It takes 2 hours to load a truck.
- Truck B is in the yard now. It takes 2 hours to load.
- Only one truck can be loaded at the loading dock at a time.
- Current time is 11:00."""

response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

print(response.text)
````

# Modular prompting

````python
prompt = """You are an expert Python backend developer. Construct a script based on the modules below.

<CONTEXT>
We are migrating our legacy user database to a new PostgreSQL instance. We need a script to fetch user profiles from the old REST API and insert them into the new DB.
</CONTEXT>

<TASK>
Write a Python script using `asyncio` and `aiohttp` to fetch the data, and `asyncpg` for the database insertion.
</TASK>

<CONSTRAINTS>
1. You must implement a rate limiter of max 10 requests per second.
2. Passwords must be hashed using bcrypt before insertion.
3. Use environment variables for all secrets; never hardcode credentials.
4. Implement a retry mechanism with exponential backoff for failed API calls.
</CONSTRAINTS>

<OUTPUT_FORMAT>
Return ONLY valid, well-commented Python code. Do not include markdown explanations before or after the code block.
</OUTPUT_FORMAT>"""

response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

print(response.text)
````
