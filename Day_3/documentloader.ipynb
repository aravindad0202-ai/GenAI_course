# Text loader

```python
from langchain_community.document_loaders import TextLoader

def load_text_file(file_path):
    loader = TextLoader(file_path)
    documents = loader.load()
    
    print(f"Loaded {len(documents)} document(s) from Text file.")
    print(documents[0].page_content)
    return documents

load_text_file(r'D:\IT\AI\teaching\GenAI_course\Day_3\docs\LLM_Comprehensive_Guide.txt')
```

# pdf loader

```python
from langchain_community.document_loaders import PyPDFLoader


def load_pdf_file(file_path):
    loader = PyPDFLoader(file_path)
    documents = loader.load()
    
    print(f"Loaded {len(documents)} page(s) from PDF.")
    for doc in documents:
        print(f"Page: {doc.metadata.get('page')}")
        print(doc.page_content)
        print("-" * 20)
        
    return documents

load_pdf_file(r"D:\IT\AI\teaching\GenAI_course\Day_3\docs\LLM_Comprehensive_Guide.pdf")
```

# Web base loader

```python
from langchain_community.document_loaders import WebBaseLoader

def load_web_page(url):
    print(f"Loading content from: {url}")
    # Initialize the loader with the URL
    loader = WebBaseLoader(url)
    
    # Load the document
    documents = loader.load()
    
    print(f"Loaded {len(documents)} document(s).")
    
    # Print the first 200 characters of the extracted text
    if documents:
        print("\n--- Extracted Content Snippet ---")
        print(documents[0].page_content.strip())
        print("---------------------------------")
        print("Metadata:", documents[0].metadata)
        
    return documents

# Example usage:
docs = load_web_page("https://en.wikipedia.org/wiki/Large_language_model")
```

```python
from langchain_community.document_loaders import WebBaseLoader

def load_multiple_pages(urls):
    print(f"Loading {len(urls)} web pages...")
    loader = WebBaseLoader(urls)
    
    # Load all documents
    documents = loader.load()
    
    for i, doc in enumerate(documents):
        source = doc.metadata.get('source', 'Unknown source')
        print(f"\nDocument {i+1} from {source}:")
        print(doc.page_content[:150].strip() + "...")
        
    return documents


url_list = [
    "https://en.wikipedia.org/wiki/Artificial_intelligence",
    "https://en.wikipedia.org/wiki/Machine_learning"
]
multi_docs = load_multiple_pages(url_list)
```

# Meta data enrichment

```python
import time
from langchain_core.documents import Document
from langchain_community.document_loaders import TextLoader

def add_metadata_to_documents(file_path):

    # Load the document
    loader = TextLoader(file_path)
    documents = loader.load()
    
    print("Original Metadata:", documents[0].metadata)
    
    # 3. Add new metadata to the loaded documents
    for doc in documents:
        # doc.metadata is just a Python dictionary, so we can add or update keys easily
        doc.metadata["author"] = "Jane Doe"
        doc.metadata["category"] = "Finance"
        doc.metadata["classification"] = "Confidential"
        doc.metadata["processed_timestamp"] = time.time()
        
    print("\nUpdated Metadata:", documents[0].metadata)
    
    return documents

add_metadata_to_documents(r'D:\IT\AI\teaching\GenAI_course\Day_3\docs\LLM_Comprehensive_Guide.txt')
```
